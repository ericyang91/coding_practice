## Linear Regression

### Linear Regression Basics

**Linear Regression**

- Linear Regression models propose a linear relationship between a collection of independent variables ($X$) and a single dependent variable ($Y$).
- Linear Regression modelling attempts to find the "best" relationship between a single dependent variable ($Y$) and a collection of one or more explanatory variables ($x_1, x_2, \dots, x_k$).
- $$y = \beta_0 + \beta_1 x_1 + \dots + \beta_k x_k + e$$
- The "linear" in linear regression does not mean that these models can only capture straight lines
- Linear Regression can capture the impact of a categorial/discrete (dummy) variable on either the level of the dependent variable, the marginal impact of an independent variable (or both!)
- Regression has two primary uses:
- 1. Predictions/ Forecasts
  2. Understanding variation

**Categorical Variables in LR**

- Linear regression models can capture the impact of discrete or categorical variables—often represented as binary dummy variables, where $D \in \{0, 1\}$—in two primary ways: shifting the baseline level or shifting the marginal impact.

1. **Shifting the Level (The Dummy Variable):** This approach models a baseline difference between two groups. The primary independent variable affects both groups at the exact same rate, resulting in parallel regression lines.

  - The General Equation: $$y = \beta_0 + \beta_1 x_1 + \beta_2 D$$

  - Base Group ($D = 0$): The dummy term zeroes out. The equation collapses to $y = \beta_0 + \beta_1 x_1$. The y-intercept is strictly $\beta_0$.
  - Shifted Group ($D = 1$): The equation becomes $y = (\beta_0 + \beta_2) + \beta_1 x_1$. The intercept shifts up or down by the exact value of $\beta_2$.

  - Practical Example:
    - Predicting base salary ($y$) based on years of experience ($x_1$) and holding a graduate degree ($D$).
      - Every year of experience adds $\$5,000$ to the salary (the slope, $\beta_1$).
      - A graduate degree provides a flat $\$15,000$ premium (the level shift, $\beta_2$).
      - Result: Both degree holders and non-degree holders see their salaries rise at $\$5,000$ per year, but degree holders start on a parallel trend line that is $\$15,000$ higher on y-axis.

2. **Shifting the Marginal Impact (The Interaction Term):** This approach models situations where the independent variable has a distinctly different rate of effect (slope) depending on the category. The baseline intercept remains identical, but the trajectories diverge.

  - The General Equation: $$y = \beta_0 + \beta_1 x_1 + \beta_2 x_1 D$$

  - Base Group ($D = 0$): The interaction term zeroes out. The equation is $y = \beta_0 + \beta_1 x_1$, yielding a standard slope of $\beta_1$.
  - Shifted Group ($D = 1$): The equation factors to $y = \beta_0 + (\beta_1 + \beta_2)x_1$. The slope changes by the magnitude of $\beta_2$.

  - Practical Example:
    - Predicting total goals scored ($y$) based on minutes played ($x_1$) and playing as a forward ($D$).
      - A non-forward scores $1$ goal per $1,000$ minutes (the base slope, $\beta_1$).
      - A forward scores an *additional* $4$ goals per $1,000$ minutes (the marginal shift, $\beta_2$).
      - Result: Both player types start at exactly $0$ goals at $0$ minutes (shared intercept, $\beta_0$). However, the trend line for forwards climbs much more steeply as minutes accumulate, reflecting a higher marginal return on time played.

**Missing Observations**

1. Missing Complete at Random (MCAR):
- The probability of an observation being missing has absolutely nothing to do with its own value or any other values in your dataset. It is a purely random anomaly.
- A network timeout randomly drops a handful of transaction records during a scheduled ETL transfer.
- You can use Multiple Imputation (MI), although it's less effective.
- Preferred to delete the observation.

2. Missing at Random (MAR):
- The missingness is systematically related to other observed variables in your dataset, but not to the missing value itself.
- When processing demographic data, younger users might be less likely to fill out an "annual income" field. The missingness is tied to their age (which you have), but within that specific age bracket, the missingness doesn't depend on exactly how much money they actually make.
- You can use Multiple Imputation (MI)

3. MNAR (Missing Not at Random): 
- The most problematic scenario. The missingness depends directly on the unobserved value itself.
- Individuals with exceptionally high financial leverage deliberately leave a "total debt" field blank because the number is so high.
- If observations are not missing at random, the consequence of dropping them depends on the **nature of that pattern.** While you might still drop the observations and caveat the results, you should consider some form of imputation to address the problem. Any process you use must model the pattern of missingness to not introduce bias. In this sense, MI can be used, but the data provided must contain the pattern of missingness Five methods come to mind here:
- Prior knowledge
- Worst case scenario 
- Mean substitution
- Regression with multiple imputation (more on this later)
- Full information maximum likelihood (FIML)

**Outliers**

Outliers are observations that deviate significantly from the distribution of the remaining observations. They tend to be very impactful on many types of analysis. These observations belong to one of the following categories.
1. An error -> correct, impute or delete
2. An extreme value that is out of scope -> delete
3. An extreme value that is in scope -> retain, but with caution
Note: there are methods of managing outliers that reduce their impact on analysis. Be cautious when using transformations or recoding to reduce the impact of outliers.

Outliers could be univariate or multivariate. 
1. Univariate: Unconditionally extreme value; Easily detectable via summary statistics, boxplots, etc.
2. Multivariate: Conditionally extreme value; Often detected after-the-fact using influence calculations. 

**Truncated/ Censored Data**

A sample is truncated when observations at tail ends of a distribution are removed, such as very high-income earners or sufficiently late deliveries. This can create misleading results because the extreme values have no impact on the model. A sample is censored when records with extreme values are retained but the extreme value is recoded as a different value – typically a boundary value. Methods exist for working with truncated or censored data.

### Testing a Linear Regression Model

Work through these in order, and stop when one fails:

1. **Does the model have any predictive power?** Check the joint F-test.
2. **Do the variables in the model belong there?** Check the t-tests.
3. **Have we violated regression assumptions?** Check the four diagnostic plots.
4. **Is the R² (or adjusted R²) sufficient for the business need?** This is a judgment call, not a threshold.

#### Joint F-Test

First, we ask the global question 'is this regression model any good at all?' To do so, we use the **Joint F-Test** - a formal hypothesis test to assess whether a model has any predictive ability. It tests the null hypothesis that all coefficients are jointly equal to $0$. 

The hypotheses are:

$$H_0 : \beta_1 = \beta_2 = ... = \beta_k = 0$$

$$H_1 : \text{at least one } \beta_i \neq 0$$

$$F = \text{MSR}/\text{MSE} \sim F_a(ndf = k, ddf = n - k - 1)$$

Where:

$$\text{MSR} = \text{Mean Squared Regression} = \text{SSR}/k$$

$$\text{MSE} = \text{Mean Squared Error} = \text{SSE}/(n - k - 1)$$

Example (Is the team any good?):

Before you look at individual players, you need to know if the team as a whole is actually capable of winning. The Math: The F-Test asks, "Are all of your multipliers ($\beta_1, \beta_2, \beta_3$) basically just zero?" (meaning they have absolutely no effect). The Business Meaning: It tests if your entire model is completely useless. If your model fails the F-Test, it means your predictions are no better than just guessing the average Order_Size every single time. If it passes, it proves your model as a whole actually has real predictive power.

#### T-Tests

If the model passes the joint-F, next we test individual coefficient terms using a **t-test**. Each test assesses whether the coefficient for an individual variable is equal to $0$. The hypotheses here will be of the form:

$$H_0 : \beta_1 = 0$$

$$H_1 : \beta_1 \neq 0$$

$$t = \frac{b_1 - \beta_1}{S_{b1}} \sim t_{df=n-k-1}$$

Where:

$$Se = \sqrt{\frac{SSE}{n-k-1}} \qquad S_{b1} = \frac{Se}{\sqrt{(n-1)S_x^2}}$$

Example: The t-tests (Is this specific player any good?):

If the team is good (you passed the F-Test), you now want to evaluate the individual players. The Math: A t-test looks at just one specific variable (like Ad_Budget) and asks, "Is this specific multiplier ($\beta_1$) actually zero?" The Business Meaning: Let's say your model passed the F-Test, so you know it works. But when you look at the t-tests, Ad_Budget passes, but Distance fails. This tells the manager: "Our advertising budget is definitively driving sales. But distance? The data says distance is just random noise and isn't actually affecting our order sizes. We should remove distance from our strategy."

In statistical terms, if the p-value of a test-statistic is less than alpha, we reject the null hypothesis that the coefficient of the independent variable is equal to zero, we embrace the alternative hypothesis that the coefficient is not equal to zero, and we conclude that the independent variable is a statistically significant predictor of the dependent variable.

#### R-Square

$R^2$ is an assessment of model fit – it represents the proportion of variability in $y$ explained by the model. The **Adjusted** $R^2$ adjusts for the number of variables included in a model, and is a better measure of model fit. $R^2$ increases with the number of factors.**Adjusted** $R^2$ increases or decreases depending on whether adding a factor adds enough accruacy to overcome the penalty.

In some contexts, a regression may only be useful if it explains a significant amount of the variation. There are several measures of this, the most important are R2 (and later, adjusted R2.)

$$R^2 = 1 - \frac{SSE}{SST} = \frac{SST - SSE}{SST} = \frac{SSR}{SST} \leq 1$$

Where:

$$\text{SSE} = \text{Sum of Squares Error} = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2 = \sum_{i=1}^{n}\hat{\epsilon}_i^2$$

$$\text{SSR} = \text{Sum of Squares Regression} = \sum_{i=1}^{n}(\hat{y}_i - \bar{y}_i)^2$$

$$\text{SST} = \text{Sum of Squares Total} = \sum_{i=1}^{n}(y_i - \bar{y}_i)^2$$

### Assessing a Linear Regression Model

Assessing involves:
- **Checking for the distribution of residuals:** Residuals are the set of differences between our training data and our model's prediction. A big indicator in the efficacy of the regression model is whether or not the residuals follow a **Normal Distribution**. This would suggest that the errors are due to random error and not some systematic flaw in our model. There are a few plots we can use to check this.
- **Error conditions**: Check if errors are normally distributed, mean of the distribution is 0, with constant standard deviation (homoskedasticity)
    - **Heteroskedasticity**: The variance of the residuals (the error terms) is not constant across all observations. Your errors shouldn't get wider as your numbers get bigger. For example, your model shouldn't be highly accurate for small orders but wildly inaccurate for massive orders. The "spread" of your errors should stay relatively constant. In heteroskedasticity, errors increase or decrease as the predicted Y (dependent variable) changes. For instance, It might predict small customer orders perfectly (smaller errors), but when a giant corporate customer comes in, its predictions are off by hundreds or thousands of dollars. Heteroskedasticity is generally not as serious a problem, it means that your estimates are inefficient and your hypothesis tests are wrong, but these can be corrected with Generalized Least Squares (GLS) or (for the hypothesis tests) robust estimation techniques.
        - We can assess the distribution of the residuals using a **histogram**. A non-normal error term may not be a big problem. It will impact confidence intervals and hypothesis tests but can generally be addressed with more sophisticated techniques.
            
- **Outliers/influence**: As discussed above
- **Sign direction**: Check the magnitude and sign direction of the coefficients
- **Error must be independent**
    - **Autocorrelation**: Error is not independent, rather it is related to previous values of itself. This only occurs with time series data; This can be a very serious problem with time series data. In some cases, it can make the results meaningless.

#### Normal Q-Q (Quantile-Quantile)

```python
import scipy.stats as stats

'''
Boilerplate code for creating a Normal Q-Q (Quantile-Quantile) plot, which is used to test normality. 
The first two lines declare a figure and a subplot. This is an alternate way to output plots which allows 
for more than one plot per output. Specifically, this plot is checking the residuals of your regression model against 
a theoretical normal distribution.
The blue dots represent the actual distribution of your data (the ordered residuals), 
and the red diagonal line represents what a perfectly normal distribution would look like.
If the blue dots lie above the red line, it means that the errors (residuals) are higher than expected, meaning
the model under-predicted the actual value. And vice versa.
Theoretical quantiles indicate the number of standard deviations away from the mean.
'''
fig = plt.figure()
ax = fig.add_subplot(111)

'''
Scipy.Stats has a built-in function for generating this type of plot. This 
function takes three arguments: the measurement being checked (residuals), 
the distribution we are checking against (normal in this case), and the plot 
to plot it to.
'''
stats.probplot(residuals,dist='norm',plot=ax)
plt.show()
```

#### Scale-Location

Use this plot to check the assumptions of equal variance in errors (**homoscedasticity**). You want to see a horizontal line with points spread randomly on either side. If there is a clear cone shaped pattern, it is evidence of heteroskedasticity.

This plot has the predicted values on the x axis and the square root of the absolute value of the standardized residuals on the y axis. When a set of variables become standardized this means that they are referred to by the number of standard deviations that they lie away from their mean.

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

# First we create an array of normalized residuals using a Scikit function
scaler = StandardScaler().fit(residuals.reshape(-1,1))
norm_residuals = scaler.transform(residuals.reshape(-1,1))

# Plot and take the root and absolute values of the norms
plt.scatter(predicted_y,np.sqrt(np.abs(norm_residuals)),c='black',s=2)
plt.xlabel("Fitted Values")
plt.ylabel("Root of standardized residual")
plt.show()
```

#### Density Plot

There is one last plot worth observing – the density plot. This is another plot used to check for **normality of errors**. You would like to see a largely normal distribution – any other distinct pattern is indicative of a violation in assumptions. You can view it with the following code:

```python
from scipy.stats import norm

# Fit a normal distribution to the data:
mean, std = norm.fit(residuals)

# Plot the histogram.
plt.hist(residuals, bins=13, edgecolor='black', density=True)

# Generate a PDF based on the fitted distribution
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mean, std)
plt.plot(x, p, color='black')
title = "Fit results: mu = %.2f,  std = %.2f" % (mean, std)
plt.xlabel("Residuals")
plt.ylabel("Probability Density")
plt.title(title)

plt.show()
```

#### The Cook's Distance Plot (Residual vs. Leverage)

The Cook’s Distance plot helps you identify observations with **high leverage** (i.e., those that greatly influence the regression results). In this plot, the x axis is leverage, the y axis is the Studentized residuals (this is similar to standardized residual) and the size of the points indicates the Cook's Distance. Such a plot can be generated with the ${influence\_plot}$ function from the **StatsModel** library. We use the previously trained model in the code below.

1. The Seesaw Analogy (The Physical View)
    - Imagine your regression line (your model) is a seesaw, and your data points are people sitting on it.

    - Leverage (The X-Axis) is where you sit: If you sit right near the middle fulcrum, you don't have much leverage. If you sit way out at the very end of the seesaw, you have High Leverage. You are in a position where you have the potential to tilt the whole thing easily.

    - Residual (The Y-Axis) is how heavy you are: If you are exactly the weight everyone expects, you have no residual. If you drop a massive weight on your lap, you have a high residual.

    - Cook's Distance (Circle Size) is how much the seesaw actually tilted:

    - If you sit at the very edge (High Leverage) but you weigh exactly what is expected of you, the seesaw stays balanced. You had the potential to move it, but you didn't. (Small Circle).

    - If you sit perfectly in the middle, but drop a massive anvil on your lap (High Residual), the seesaw is going to bend and groan. (Large Circle).

```python
#In a similar fashion to how the QQ plot was built, this plot can be generated.
fig, ax = plt.subplots(figsize=(12,8))
fig = sm.graphics.influence_plot(model, ax=ax, criterion="cooks")
plt.show()
```

```python
"""
a.	Import the data from tab ‘Bimodal Error 1’.
b.	Run the linear regression model y = b0 + b1 x1 + b2 x2 with the default settings.
c.	Examine the regression output table.  Does it look OK?  
d.	Now look at the distribution of the error term.  What problems do you see here?
"""

# Define the path and the DataFrame
path = os.path.join(os.path.curdir, "MMA 860 Assessing and Testing Data File v1.0.xlsx")
df = pd.read_excel(path, sheet_name="Bimodal Error 1", index_col="Obs")

# Train the model and print the summary
model = ols('Y ~ X1 + X2', df).fit()
print(model.summary())

print(f"\nAbove model returns a p-value for F-statistic of 4.30e-206, less than the defined alpha level of 0.05\nindicating a significant predictive power of the model. The adjusted R-squared is 0.992\nmeaning that 99.2% of the variance of Y is explained by X1 and X2.\nWhen we look at the p-values of the individual t-tests, we see that it is 0.0001 for both X1 and X2.")
```

```python
# Train the data
train_X = df[['X1','X2']].values
train_y = df['Y'].values

reg = LinearRegression().fit(train_X, train_y)

# Residuals calculated by definition above.
predicted_y = reg.predict(train_X)
residuals = train_y - predicted_y

# Fit a normal distribution to the data:
mean, std = norm.fit(residuals)

# Plot the histogram.
plt.hist(residuals, bins=13, edgecolor='black', density=True)

# Generate a PDF based on the fitted distribution
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mean, std)
plt.plot(x, p, color='black')
title = "Fit results: mu = %.2f,  std = %.2f" % (mean, std)
plt.xlabel("Residuals")
plt.ylabel("Probability Density")
plt.title(title)

plt.show()

print(f"\n\nWe can see a bimodal distribution of the residuals. Probability density of the residuals peak at around -200 and 300.\nThe model is rarely making small prediction errors, but is consistently making large negative or positive errors.\nIn a healthy model, the distribution should be normally distributed, meaning it should not make,\nor make very small errors most of the time, and make very few large errors.\nBecause the F-test and T-tests assume that the distribution of the residuals follows a bell curve,\nwe can no longer trust the validity of these tests.\nWhen residuals split into two separate peaks like this, it is a massive red flag\nthat your model is fundamentally missing a piece of the puzzle.\nIt usually means there are two distinct sub-populations hidden within your data,\nand your model is trying to draw a single line right down the middle of them. Because the line splits\nthe difference, it consistently over-predicts one group (causing the negative peak) and under-predicts the other group (causing the positive peak).")
```

### Modelling Approaches

- Modelling Philosophies
- 1. Average Econometric Regression (AER)
- 2. Test Test Test (TTT)
- 3. Fragility Analysis

- Model Development General Rules:
1. Models should start with a theory
2. Data used to develop the model should not be used to test it
3. Testing down generally creates less bias
4. Multiple tests create compounding errors, consider using singular tests of multiple constraints where feasible
5. Models should always be subject to specification and other tests before being accepted
